```python
import kaggle
import pandas as pd
import os

kaggle.api.authenticate()

# kaggle.api.dataset_download_files('paradisejoy/top-hits-spotify-from-20002019', path='../data/raw/', unzip=True)
if os.path.exists("../data/raw/SpotifyRaw.csv"):
    os.remove("../data/raw/SpotifyRaw.csv")
# if os.path.exists("../data/raw/songs_normalize.csv"):
#     os.rename("../data/raw/songs_normalize.csv","../data/raw/SpotifyRaw.csv")
# df_spotify = pd.read_csv("../data/raw/SpotifyRaw.csv")
df_spotify = pd.read_csv("hf://datasets/maharshipandya/spotify-tracks-dataset/dataset.csv")
df_spotify.to_csv("../data/raw/SpotifyRaw.csv")


kaggle.api.dataset_download_files("sansastark/subset-of-the-million-song-dataset", path='../data/raw/', unzip=True)
if os.path.exists("../data/raw/billboard_rank.csv"):
    os.remove("../data/raw/billboard_rank.csv")
if os.path.exists("../data/raw/MillionSongSubset.csv"):
    if os.path.exists("../data/raw/MillionSongRaw.csv"):
        os.remove("../data/raw/MillionSongRaw.csv")
    os.rename("../data/raw/MillionSongSubset.csv","../data/raw/MillionSongRaw.csv")
df_million_song = pd.read_csv("../data/raw/MillionSongRaw.csv")
```

```python
# Standardize name columns
df_spotify= df_spotify.drop_duplicates()
df_spotify = df_spotify[df_spotify["popularity"] != 0]
df_spotify = df_spotify.drop_duplicates(subset=['artists','track_name'], keep='last')
df_spotify['track_name_clean'] = df_spotify['track_name'].str.lower().str.strip()
df_spotify['artist_clean'] = df_spotify['artists'].str.lower().str.strip()

df_million_song = df_million_song.sort_values("Year").drop_duplicates(subset=["ArtistName", "Title"], keep='last')
df_million_song['song_clean'] = df_million_song['Title'].str.lower().str.strip()
df_million_song['artist_clean'] = df_million_song['ArtistName'].str.lower().str.strip()

merged_exact = df_spotify.merge(
    df_million_song,
    left_on=['artist_clean', 'track_name_clean'],
    right_on=['artist_clean', 'song_clean'],
    how='inner'
)

print(len(merged_exact))
```
