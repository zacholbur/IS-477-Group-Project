### Description of Cleaning Process

#### Spotify Dataset
The following actions were taken to clean the spotify data in response to the data quality assessment we conducted:
- **Popularity Scores of 0:** Given that our analysis is centered around the popularity field, fields with a popularity of 0, which is clearly inaccurate, are not useful for our analysis. As a result, we removed all such observations
- **Drop Complete Duplicates**: Next, we removed all observation that were complete duplicates of another observation.
- **Handle Other Duplicates**: Finally, we needed to handle effective duplicates, which we defined as observations that had the same song and artist.We decided to use a rule-based method of keeping the more recent entry and dropping the less recent entry.

*Below, please find our code for conducting the outlined data cleaning on the spotify dataset*

```python
import pandas as pd
df_spotify = pd.read_csv("../data/raw/SpotifyRaw.csv")

df_spotify["ArtistLower"] = df_spotify["artists"].str.lower()
df_spotify["SongLower"] = df_spotify["track_name"].str.lower()


df_spotify = df_spotify[df_spotify["popularity"] != 0]
df_spotify = df_spotify.drop_duplicates()
df_spotify = df_spotify.drop_duplicates(subset=['ArtistLower','SongLower'], keep='last')

df_spotify.to_csv("../data/processed/SpotifyCleaned.csv")
```

#### Million Song Dataset
The follow actions were taken to clean the million song data in response to the data quality assessment we conducted:
- **Remove Columns with High Number of Missing Values**: As outlined in our quality assessment, the ArtistLongitude ArtistLocation, and ArtistLatitude fields have a large amount of missing data. Similarly, the Danceability and Energy fields are entirely made up of 0 columns and are effectively missing. With thin in mind, we removed all 5 of these columns.
- **Drop Complete Duplicates**: Next, we removed all observation that were complete duplicates of another observation.
- **Handle Other Duplicates**: We also needed to handle effective duplicates, which we defined as observations that had the same song and artist. We decided to use a rule-based method of keeping the more recent entry and dropping the less recent entry. This works particularly well for this dataset because we sorted by year. Furthermore, any entries with a year of 0 would be given least priority in this method.
- **Remove Year Column**: Finally, after it had been used to sort, we removed the year column because of the high number of missing values.

*Below, please find our code for conducting the outlined data cleaning on the million song dataset*

```python
df_million_song = pd.read_csv("../data/raw/MillionSongRaw.csv")

df_million_song["ArtistLower"] = df_million_song["ArtistName"].str.lower()
df_million_song["SongLower"] = df_million_song["Title"].str.lower()

df_million_song = df_million_song.drop(["ArtistLongitude", "ArtistLocation","ArtistLatitude","Danceability", "Energy"], axis=1)
df_million_song = df_million_song.sort_values("Year").drop_duplicates(subset=["ArtistLower", "SongLower"], keep='last')
df_million_song = df_million_song.drop(["Year"], axis=1)

df_million_song.to_csv("../data/processed/MillionSongCleaned.csv")
```
